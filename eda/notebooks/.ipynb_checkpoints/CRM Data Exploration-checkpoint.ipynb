```python
from pathlib import Path
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

pd.set_option("display.max_columns", 50)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", lambda x: f"{x:,.3f}")

# Expected repository layout:
# 2Dra0tHBwdCEpOhD/
# ├── data/                 # local only; add to .gitignore
# │   └── SalesCRM.xlsx
# └── notebooks/
#     └── 01_data_exploration.ipynb

CANDIDATE_PATHS = [
    Path("SalesCRM.xlsx")
    
]

DATA_PATH = next((p for p in CANDIDATE_PATHS if p.exists()), None)

if DATA_PATH is None:
    raise FileNotFoundError(
        "CRM workbook not found. Place it in the project's data/ folder "
        "as data/SalesCRM.xlsx, or update DATA_PATH to its local path."
    )

# Read the CRM sheet. Excel may contain malformed date-formatted cells;
# pandas/openpyxl can warn and treat those individual cells as missing.
with warnings.catch_warnings():
    warnings.simplefilter("ignore", UserWarning)
    raw_df = pd.read_excel(DATA_PATH, sheet_name=0)

df = raw_df.copy()

print(f"Loaded workbook: {DATA_PATH.name}")
print(f"Dataset shape: {df.shape[0]:,} rows × {df.shape[1]:,} columns")
display(df.head())
```

```python
print("Column names:")
for column in df.columns:
    print(f" • {column}")

print("\\nData types:")
display(df.dtypes.rename("dtype").to_frame())

print("\\nDataset information:")
df.info()
```

```python
missing_summary = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_pct": df.isna().mean() * 100,
    "non_missing_count": df.notna().sum(),
    "unique_non_missing": df.nunique(dropna=True),
}).sort_values("missing_pct", ascending=False)

display(missing_summary.round(2))

ax = missing_summary["missing_pct"].sort_values().plot(
    kind="bar", figsize=(12, 5), ylim=(0, 100)
)
ax.set_title("Missing values by CRM field")
ax.set_ylabel("Missing values (%)")
ax.set_xlabel("")
ax.tick_params(axis="x", rotation=55)
plt.tight_layout()
plt.show()
```

```python
print("Duplicate full rows:", int(df.duplicated().sum()))

if "ID" in df.columns:
    print("Missing IDs:", int(df["ID"].isna().sum()))
    print("Duplicate IDs:", int(df["ID"].duplicated().sum()))
    print("Unique IDs:", int(df["ID"].nunique(dropna=True)))

# ID is an identifier, not a customer attribute. Exclude it from model features.
```

```python
categorical_candidates = [
    c for c in ["Country", "Education", "Status", "Stage"]
    if c in df.columns
]

for column in categorical_candidates:
    print(f"\\n{column}: {df[column].nunique(dropna=True)} unique non-missing values")
    display(
        df[column].value_counts(dropna=False)
          .head(20)
          .rename_axis(column)
          .reset_index(name="records")
    )
```

```python
# Plot the most common values for the main categorical fields.
for column in categorical_candidates:
    counts = df[column].fillna("Missing").astype(str).value_counts().head(15)
    fig, ax = plt.subplots(figsize=(9, 4))
    counts.sort_values().plot(kind="barh", ax=ax)
    ax.set_title(f"Top values — {column}")
    ax.set_xlabel("Number of records")
    ax.set_ylabel("")
    plt.tight_layout()
    plt.show()
```

```python
date_like_columns = [
    "First Contact",
    "Last Contact",
    "First Call",
    "Signed up for a demo",
    "Filled in customer survey",
    "Did sign up to the platform",
    "Account Manager assigned",
    "Subscribed",
]

# Convert values to datetimes safely. Non-date values (for example, "No")
# become NaT in the parsed helper series but remain untouched in df.
parsed_dates = {}
for column in date_like_columns:
    if column in df.columns:
        parsed_dates[column] = pd.to_datetime(df[column], errors="coerce")

date_summary_rows = []
for column, series in parsed_dates.items():
    date_summary_rows.append({
        "field": column,
        "parseable_dates": int(series.notna().sum()),
        "earliest_date": series.min(),
        "latest_date": series.max(),
        "raw_non_missing": int(df[column].notna().sum()),
    })

date_summary = pd.DataFrame(date_summary_rows).set_index("field")
display(date_summary)
```

```python
# Count recorded values for every funnel milestone.
# For event columns, a non-empty value is evidence of a recorded value,
# not necessarily proof of successful completion unless the field semantics confirm it.
milestone_columns = [
    "First Call",
    "Signed up for a demo",
    "Filled in customer survey",
    "Did sign up to the platform",
    "Account Manager assigned",
    "Subscribed",
]
milestone_columns = [c for c in milestone_columns if c in df.columns]

milestone_summary = pd.DataFrame({
    "raw_non_missing": df[milestone_columns].notna().sum(),
    "recorded_pct": df[milestone_columns].notna().mean() * 100,
}).sort_values("raw_non_missing", ascending=False)

display(milestone_summary.round(2))

fig, ax = plt.subplots(figsize=(10, 5))
milestone_summary["raw_non_missing"].sort_values().plot(kind="barh", ax=ax)
ax.set_title("Records with a non-missing acquisition milestone")
ax.set_xlabel("Number of records")
ax.set_ylabel("")
plt.tight_layout()
plt.show()
```

```python
if "Subscribed" in df.columns:
    subscribed_recorded = df["Subscribed"].notna()
    n_total = len(df)
    n_recorded = int(subscribed_recorded.sum())
    n_missing = int(df["Subscribed"].isna().sum())

    print(f"Records: {n_total:,}")
    print(f"Subscribed field populated: {n_recorded:,} ({n_recorded / n_total:.2%})")
    print(f"Subscribed field missing: {n_missing:,} ({n_missing / n_total:.2%})")

    print("\\nRaw value examples (aggregate counts only):")
    display(df["Subscribed"].value_counts(dropna=False).head(10).rename("records").to_frame())
else:
    print("Subscribed field is not present.")
```

```python
# These are descriptive conversion rates only.
# They do not adjust for confounding, exposure time, or selection bias.
if "Subscribed" in df.columns:
    eda_df = df.copy()
    eda_df["subscription_recorded"] = eda_df["Subscribed"].notna().astype(int)

    for segment in ["Country", "Education", "Status", "Stage"]:
        if segment not in eda_df.columns:
            continue

        summary = (
            eda_df.groupby(segment, dropna=False)
            .agg(
                prospects=("subscription_recorded", "size"),
                recorded_subscriptions=("subscription_recorded", "sum"),
                observed_rate=("subscription_recorded", "mean"),
            )
            .sort_values(["prospects", "observed_rate"], ascending=[False, False])
        )
        summary["observed_rate_pct"] = summary["observed_rate"] * 100

        print(f"\\n{segment}: largest groups (minimum group size not applied)")
        display(summary.head(20).drop(columns="observed_rate").round(2))
```

```python
# Visualize only the most frequent categories, and show sample size alongside rate.
if "Subscribed" in df.columns:
    plot_df = df.copy()
    plot_df["subscription_recorded"] = plot_df["Subscribed"].notna().astype(int)

    for segment in ["Country", "Education"]:
        if segment not in plot_df.columns:
            continue
        summary = (
            plot_df.groupby(segment, dropna=False)
            .agg(prospects=("subscription_recorded", "size"),
                 observed_rate=("subscription_recorded", "mean"))
            .sort_values("prospects", ascending=False)
            .head(12)
        )
        summary = summary.sort_values("observed_rate")
        labels = summary.index.astype(str)

        fig, ax = plt.subplots(figsize=(9, 5))
        ax.barh(labels, summary["observed_rate"] * 100)
        ax.set_title(f"Observed subscription-record rate by {segment}")
        ax.set_xlabel("Records with Subscribed populated (%)")
        ax.set_ylabel("")
        for y, (_, row) in enumerate(summary.iterrows()):
            ax.text(row["observed_rate"] * 100, y,
                    f"  n={int(row['prospects']):,}", va="center", fontsize=8)
        plt.tight_layout()
        plt.show()
```

```python
if "First Call" in parsed_dates:
    first_call_dates = parsed_dates["First Call"].dropna()
    if len(first_call_dates):
        monthly_calls = first_call_dates.dt.to_period("M").value_counts().sort_index()
        monthly_calls.index = monthly_calls.index.astype(str)

        fig, ax = plt.subplots(figsize=(11, 4))
        monthly_calls.plot(kind="bar", ax=ax)
        ax.set_title("Recorded first calls by month")
        ax.set_xlabel("Month")
        ax.set_ylabel("Number of recorded first calls")
        ax.tick_params(axis="x", rotation=60)
        plt.tight_layout()
        plt.show()
    else:
        print("No parseable First Call dates were found.")
```

```python
OUTPUT_DIR = Path("../eda/")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

missing_summary.to_csv(OUTPUT_DIR / "missingness_summary.csv")

if "milestone_summary" in globals():
    milestone_summary.to_csv(OUTPUT_DIR / "milestone_summary.csv")

print(f"Aggregate summaries saved to: {OUTPUT_DIR.resolve()}")
```
